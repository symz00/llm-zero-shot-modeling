```python
import pandas as pd
```

```python
def generate_datasets(dataset_orig_name, column_names, dataset_save_name):
    df = pd.read_csv(
        f"./data/Olympus/dataset_{dataset_orig_name}/data.csv",
        header=None,
        names=column_names,
    )
    print(df.info())
    display(df.head(5))

    df.to_csv(f"./results/Olympus/dataset_{dataset_save_name}.csv", index=None)
```

```python
generate_datasets(
    dataset_orig_name="alkox",
    column_names=["catalase", "peroxidase", "alcohol_oxidase", "ph", "conversion"],
    dataset_save_name="alkox",
)
```
