```python
import pandas as pd
```

## Buchwald-Hartwig reaction

```python
# Read the original dataset
df_BH = pd.read_csv("./data/Shields_et_al_2021/aryl_amination/experiment_index.csv")
df_BH.info()
df_BH.head(5)
```

```python
# Delete the "entry" column
df_BH = df_BH.drop(columns=["entry"])
```

```python
# Remove duplicates
print(f"Before: {len(df_BH)}")
df_BH = df_BH.drop_duplicates()
print(f"After: {len(df_BH)}")
```

```python
# Unique counts for each column
for col in df_BH.columns.tolist():
    print(f"{col}", df_BH[col].nunique())
```

```python
# Unique values for "Aryl_halide_SMILES"
print(df_BH["Aryl_halide_SMILES"].unique())
```

```python
# Create subsets based on "Aryl_halide_SMILES"
aryl_halide_SMILES_subsets = [
    [
        "FC(F)(F)c1ccc(Cl)cc1",
        "FC(F)(F)c1ccc(Br)cc1",
        "FC(F)(F)c1ccc(I)cc1",
    ],
    [
        "COc1ccc(Cl)cc1",
        "COc1ccc(Br)cc1",
        "COc1ccc(I)cc1",
    ],
    [
        "CCc1ccc(Cl)cc1",
        "CCc1ccc(Br)cc1",
        "CCc1ccc(I)cc1",
    ],
    [
        "Clc1ccccn1",
        "Brc1ccccn1",
        "Ic1ccccn1",
    ],
    [
        "Clc1cccnc1",
        "Brc1cccnc1",
        "Ic1cccnc1",
    ],
]

df_BH_subsets = []
for i, aryl_halide_SMILES_subset in enumerate(aryl_halide_SMILES_subsets):
    df_BH_subset = df_BH[
        df_BH["Aryl_halide_SMILES"].isin(aryl_halide_SMILES_subset)
    ].reset_index(drop=True)
    print(f"Sample size of subset {i + 1}:", len(df_BH_subset))
    df_BH_subsets.append(df_BH_subset)

# Verification
print(
    "Does the sum of sample sizes match the original dataset?",
    len(df_BH) == sum([len(df_BH_subset) for df_BH_subset in df_BH_subsets]),
)
```

```python
# Unique counts for each column in each subset
for i, df_BH_subset in enumerate(df_BH_subsets):
    print(f"Subset {i + 1}:")
    for col in df_BH.columns.tolist():
        print(f"{col}", df_BH_subset[col].nunique())
    print()
```

```python
# Insert ID column and save
for i, df_BH_subset in enumerate(df_BH_subsets):
    df_BH_subset.insert(0, "ID", range(1, len(df_BH_subset) + 1))
    df_BH_subset.to_csv(
        f"./results/Shields_et_al_2021/dataset_BH_{i + 1}.csv", index=None
    )
```
